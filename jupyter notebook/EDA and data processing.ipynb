```python

df=df.drop(columns=["original_label"])
```

```python
df.info()
```

```python
df["text"].isnull().value_counts()
```

```python
df["ats_score"].isnull().value_counts()
```

```python
df['text'].str.contains(r'\bSEP\b').value_counts()
```

```python
df[['resume_text', 'job_description']] = (
    df['text'].str.split('SEP', n=1, expand=True)
)
```

```python
df.info()
```

```python
df["resume_text"]=df["resume_text"].str.split()
```

```python
df["job_description"]=df["job_description"].str.split()
```

```python
df[['resume_text', 'job_description']].head()
```

```python
df['resume_text'] = df['resume_text'].apply(' '.join)
df['job_description'] = df['job_description'].apply(' '.join)
```

```python
df[['resume_text', 'job_description']].head()
```

```python
df.info()
```

```python
df=df.drop(columns=["text"])
```

```python
df.info()
```

```python
df.to_csv("final_train_dataset.csv", index=False)
```
